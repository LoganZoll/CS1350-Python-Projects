```python
import numpy as np

temperatures = np.array([68, 72, 75, 71, 69, 74, 77])

print("Average:", temperatures.mean())
print("Highest:", temperatures.max())
print("In Celsius:", (temperatures - 32) * 5 / 9)
```

```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

plt.plot(days, temperatures, marker="o")
plt.title("Temperature This Week")
plt.xlabel("Day")
plt.ylabel("Degrees F")
plt.grid(True)
plt.show()
```

```python
plt.bar(days, temperatures, color="steelblue")
plt.title("Temperature This Week")
plt.show()
```
